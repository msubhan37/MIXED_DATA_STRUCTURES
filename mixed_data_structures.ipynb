Q41. Student Performance System
students = [
 {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
 {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
 {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]
Calculate each average, top student, tuple of names, and set of unique marks.

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]
averages = {}
unique_marks = set()
for s in students:
    averages[s["name"]] = sum(s["marks"]) / len(s["marks"])
    unique_marks.update(s["marks"])
names = tuple(s["name"] for s in students)
print(averages)
print(max(averages, key=averages.get))   
print(names)                            
print(unique_marks)
```


Q42. Employee Skills
employees = {
 "Ali": {"Python", "SQL", "AWS"},
 "Sara": {"Python", "Databricks", "SQL"},
 "Ahmed": {"AWS", "Snowflake", "SQL"}
}
Find common skills, Ali-only skills, all team skills, and employees meeting Python + SQL requirements.

```python
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}
common = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
required = {"Python", "SQL"}
qualified = []
for name, skills in employees.items():
    if required.issubset(skills):
        qualified.append(name)
print(common)      
print(ali_only)    
print(all_skills)
print(qualified) 
```

sales = [

Calculate revenue, top-revenue product, name-to-revenue dictionary, and set of prices.

courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}
Find students in exactly two, all three, and exactly one course; count unique students.

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}
p, s, a = courses["Python"], courses["SQL"], courses["AWS"]
all_three = p & s & a
exactly_two = ((p & s) | (p & a) | (s & a)) - all_three
everyone = p | s | a
exactly_one = everyone - exactly_two - all_three
print(exactly_two)  
print(all_three)      
print(exactly_one)    
print(len(everyone))
```

Q46. E-Commerce Transactions
transactions = [
 {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
 {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
 {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]
Calculate total revenue, Ali spending, largest transaction, and unique product names.

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]
total_revenue = 0
ali_spending = 0
txn_totals = {}
products = set()
for t in transactions:
    t_total = 0
    for name, price, qty in t["items"]:
        t_total += price * qty
        products.add(name)
    txn_totals[t["id"]] = t_total
    total_revenue += t_total
    if t["customer"] == "Ali":
        ali_spending += t_total
print(total_revenue)                      
print(ali_spending)                        
print(max(txn_totals, key=txn_totals.get)) 
print(products)
```

Q47. Pipeline Monitoring
pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
 "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
 "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
Find total processed records, failed pipelines, largest pipeline total, and average records per run for each.

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
total_records = 0
failed = []
p_totals = {}
p_avg = {}
for name, info in pipelines.items():
    t = sum(info["records"])
    p_totals[name] = t
    p_avg[name] = t / len(info["records"])
    total_records += t
    if info["status"] == "failed":
        failed.append(name)
print(total_records)                     
print(failed)                             
print(max(p_totals, key=p_totals.get))     
print(p_avg)
```

Q48. Banking Transactions
accounts = {
 "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
 "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
 "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
Assuming zero starting balances, calculate final balances, highest balance, total withdrawals, and largest single
deposit.

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
balances = {}
total_withdrawals = 0
largest_deposit = 0
for acc, info in accounts.items():
    balances[acc] = sum(info["transactions"])
    for amount in info["transactions"]:
        if amount < 0:
            total_withdrawals += -amount
        elif amount > largest_deposit:
            largest_deposit = amount
print(balances)                        
print(max(balances, key=balances.get))   
print(total_withdrawals)                 
print(largest_deposit)  
```

Q49. Product Recommendations
users = {
 "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
 "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
 "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
bought_count = {}
wish_count = {}
all_products = set()
for info in users.values():
    for item in info["purchased"]:
        bought_count[item] = bought_count.get(item, 0) + 1
    for item in info["wishlist"]:
        wish_count[item] = wish_count.get(item, 0) + 1
    all_products |= info["purchased"] | info["wishlist"]

bought_by_two = set()
for item, c in bought_count.items():
    if c >= 2:
        bought_by_two.add(item)
shared_wishlist = set()
for item, c in wish_count.items():
    if c >= 2:
        shared_wishlist.add(item)

print(bought_by_two)                                       
print(users["Ali"]["wishlist"] - users["Ali"]["purchased"])  
print(shared_wishlist)                                     
print(all_products)
```

Q50. Final Company Analysis
company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}
1. Payroll per department. 2. Company payroll. 3. Highest-paid employee. 4. Skills shared by all. 5. Engineering-only
skills. 6. Remaining budget per department. 7. Department with higher average salary. 8. Name-to-salary dictionary. 9.

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}
payroll = {}
avg_salary = {}
remaining = {}
dept_skills = {}
name_salary = {}
all_skill_sets = []
all_skills = set()
company_payroll = 0
total_budget = 0
for dept, info in company["departments"].items():
    dept_total = 0
    dept_skills[dept] = set()
    for name, salary, skills in info["employees"]:
        dept_total += salary
        name_salary[name] = salary
        all_skill_sets.append(skills)
        dept_skills[dept] |= skills
        all_skills |= skills
    payroll[dept] = dept_total
    avg_salary[dept] = dept_total / len(info["employees"])
    remaining[dept] = info["budget"] - dept_total
    company_payroll += dept_total
    total_budget += info["budget"]

print(payroll)                                   
print(company_payroll)                           
print(max(name_salary, key=name_salary.get))     
print(set.intersection(*all_skill_sets))         
print(dept_skills["Engineering"] - dept_skills["Analytics"])  
print(remaining)                               
print(max(avg_salary, key=avg_salary.get))      
print(name_salary)                               
print(all_skills)                                
print(total_budget >= company_payroll)
```
